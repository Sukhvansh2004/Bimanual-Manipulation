```python
from robosuite_franka_manipulation.transforms import TransformManager, StampedPose
from robosuite_franka_manipulation.environments import make_dual_arm_env

# Create environment and transform manager
env = make_dual_arm_env()
env.reset()

tf_manager = TransformManager(env)

# Print discovered frames
print("Known frames:", tf_manager.get_known_frames())

# Print frame tree
tf_manager.print_frame_tree()

# Get right EEF pose in world frame (using the correct frame name)
tf = tf_manager.get_frame_pose("Panda_right_eef", "world")
print("\nRight EEF pose in world frame:")
print(f"  Position: {tf.position}")
print(f"  Quaternion: {tf.quaternion}")

# Get camera pose in world frame
camera_tf = tf_manager.get_frame_pose("agentview", "world")
print("\nAgentview camera pose in world frame:")
print(f"  Position: {camera_tf.position}")
print(f"  Quaternion: {camera_tf.quaternion}")
```
